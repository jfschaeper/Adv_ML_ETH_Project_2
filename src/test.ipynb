```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
import pandas as pd
import numpy as np
from functions import *

# features_train = pd.read_csv('../out/features_train.csv')
# features_train = features_train.fillna(features_train.median())
X_train = pd.read_csv('../data/X_train.csv', index_col='id')
X_train_filtered = filter_signal(X_train)
y_train = pd.read_csv('../data/y_train.csv', index_col='id')

# would probably make sense to start from the first R-peak to have similar starting point


crossvalidation = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
from tslearn.neighbors import KNeighborsTimeSeriesClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

X_train, X_test, y_train, y_test = train_test_split(X_train_filtered.iloc[0:1000, :], y_train[0:1000], test_size=0.33, random_state=42)

model_baseline = KNeighborsTimeSeriesClassifier(n_neighbors=10, metric="dtw")
kNN = model_baseline.fit(X_train, y_train)
y_predict = kNN.predict(X_test)


# filtered = np.mean(cross_val_score(model_baseline, X_train_filtered, np.ravel(y_train), scoring="f1_micro", cv=crossvalidation, n_jobs=-1))
print("A filtered kNN model achieves an F1 score of: ", f1_score(y_test, y_predict))
```
